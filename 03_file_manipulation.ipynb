```python
# We will create a function to open a CSV file
import csv
# Ask the user what directory to search for CSVs in
import os
# we will define a function that is not hard coded to a csv file and can open any csv file
def print_csv(filename):
    with open(filename, "r") as csv_file:
        reader = csv.reader(csv_file)
        # this will make sure that every row is separated by tab
        for row in reader:
            print("\t".join(row))
# Test file
#print_csv("countries.csv")

directory = input("Enter directory to search: ")

# We will tell python to start this folder, look inside it and also walk through every folder inside it
#display total number of CSV files found
count = 0
for root, folders, files in os.walk(directory):
    for file in files:
        # We need to make sure that it only selects the files that end with csv
        if file.endswith(".csv"):
            # Go through every folder, look at every file
            filepath = os.path.join(root, file)
            print_csv(filepath)
            count = count + 1
print("Total CSV files found:", count)

if not os.path.isdir(directory):
    print("Directory does not exist.")
elif count == 0:
    print("No CSV files found.")
```
