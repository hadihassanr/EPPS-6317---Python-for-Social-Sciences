```python
# Import pandas and CSV files. We will first inspect 
import pandas as pd

alumni_data = pd.read_csv("alumni_data.csv")
income_debt_data = pd.read_csv("income_debt_data.csv")

# Inspect data
display(alumni_data.head())
display(alumni_data.info())
display(income_debt_data.info())
display(income_debt_data.head())
```

```python
# Read CSV without treating the first row as column names
test_data = pd.read_csv("alumni_data.csv", header=None)

test_data.head()
```

```python
# Since the alumni data CSV file has no header names, we will give it ourselves
column_names = ["id", "age", "gender", "major", "name"]

alumni_data = pd.read_csv(
    "alumni_data.csv",
    names=column_names)
# Inspect all again after adding header titles
display(alumni_data.info())
display(alumni_data.head())
display(income_debt_data.info())
display(income_debt_data.head())
```

```python
# Now we will inspect if the 'ID' column in both datasets are same before we merge
print(alumni_data['id'].head())
print(income_debt_data['ID'].head())

# Check if both datasets are the same
(alumni_data['id'] == income_debt_data['ID']).all()
```

```python
# Now that we know that both IDs are identical, we can merge
merged_data = pd.merge(alumni_data,
              income_debt_data,
              left_on="id",
              right_on="ID")
```

```python
# Now we remove ID that is duplicate from the dataset
merged_data.drop(columns=["id"], inplace=True)

merged_data.info()
merged_data.head()
```

```python
# Analysis 1, demographics - average age by major and gender with percentage
average_age_by_major = merged_data.groupby("major")["age"].mean()
average_age_by_major = average_age_by_major.round(2)
print("Average Age by Major:")
display(average_age_by_major)

# Now we will convert the count into percentages
gender_percent_by_major = (merged_data.groupby("major")["gender"].value_counts(normalize=True) * 100)
gender_percent_by_major = gender_percent_by_major.round(2)
print("Gender Percentage by Major:")
display(gender_percent_by_major)
```

```python
# Analysis 2
# Now we calculate the average income by major
average_income_by_major = (
    merged_data.groupby("major")["Income"]
        .mean().sort_values(ascending=False))
print("Average Income by Major:")

# Now we will use fstrings concept to organize the numbers
for major in average_income_by_major.index:
    print(major, f"${average_income_by_major[major]:,.2f}")
```

```python
# Analysis 3
# Now we are calculating average debt by major
average_debt_by_major = (
    merged_data.groupby("major")["Student Loan Debt"]
        .mean().sort_values())
print("Average Debt by Major:")

# Now we will clean the debt numbers using fstring
for major in average_debt_by_major.index:
    print(major, f"${average_debt_by_major[major]:,.2f}")
```

```python
# Analysis 4
# Calculate how long will it take for each student to pay off their loan
monthly_income = merged_data["Income"] / 12
monthly_payment = monthly_income * 0.08
months_to_payoff = merged_data["Student Loan Debt"] / monthly_payment

# Now we will convert the months into years
years_to_payoff = months_to_payoff / 12
# Clean the numbers to one decimal point
years_to_payoff = years_to_payoff.round(1)

# We will add this as a new column to the main dataframe
merged_data["Years to Payoff"] = years_to_payoff

# Display only the relevant columns for loan payoff analysis
merged_data[["name", "major", "Income", "Student Loan Debt", "Years to Payoff"]]
```
