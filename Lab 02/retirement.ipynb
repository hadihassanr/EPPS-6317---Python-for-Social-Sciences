## Input processing
Ask user for inputs and process them as needed before storing.


```python
# Prompt the user for the 4 inputs needed
# Get user balance & deposit input strings and convert to floating point number
balance = float(input("What is your starting balance?"))
starting_balance = balance 
deposit = float(input("Your monthly deposit amount?"))
# Get rate, convert to float, divide by 12 for monthly compounding and
# divide by 100 to convert from percentage
rate = float(input("What is your interest rate?")) / 100 / 12
# Get number of years and convert to integer
years = int(input("Over how many years?"))
```

Next we will create the fund by years table
Goals: 
* Print table title (including interest rate)
* Print column headers (Year & Fund)
* For each of the years until we reach the user-specified end, print the year number and year-end total

```python
# Calculate retirement funds in years using nested loops
print(f"Fund by years ({rate * 100 * 12:.1f}% annual interest rate)")
print("Year\tFund")
for y in range(0, years):
    for m in range(0, 12):
        balance = (balance + deposit) * (1 + rate)
    print (f"{y+1}\t${balance:,.2f}")
# Display the results
print(f"Retirement funds: ${balance:,.2f}")
```

## Millionaire Calculator
* Ask if the user wants to calculate how long it takes to become a millionaire.
* If the user answer Y:
  - Start again from the original starting balance
  - Start counting years at 0
* While the balance is below 1000,000:
* Stop once the balance reaches or exceeds 1000,000
* Print how many years it took

```python
choice = input(
    "Calculate length until you become a millionaire? (Y/N)"
)
if choice.upper() == "Y":
    millionaire_balance = starting_balance
    millionaire_years = 0
    while millionaire_balance < 1000000:
        for m in range(0, 12):
             millionaire_balance = (
                millionaire_balance + deposit
            ) * (1 + rate)
        millionaire_years += 1
    print(f"You will become a millionaire in {millionaire_years} years.")
```

## Fund by interest rate

Ask the user if they want to display the fund by interest rate table
If the user answers 'Y':
   - Print rate and column headings
   - Test annual interest rates from 1% to 30%
For each interest rate:
- Start again from the original balance
- Convert the annual interest rate to monthly rate
- Calculate the retirement fund over the selected number of years
- Print the interest rate and the final fund


```python
display_table = input(
    "Display fund by interest rate table? (Y/N)"
)

if display_table.upper() == "Y":
    print(f"Fund by interest rate ({years} years):")
    print("Rate\tFund")

    for annual_rate in range(1, 31):
        rate_balance = starting_balance
        monthly_rate = annual_rate / 100 / 12

        for y in range(0, years):
            for m in range(0, 12):
                rate_balance = (
                    rate_balance + deposit
                ) * (1 + monthly_rate)

        print(f"{annual_rate}%\t${rate_balance:,.2f}")
```
