This is where the Real Estate Calculator starts.
I added the prompt where the program is asking the user for their name as well.


```python
# We will import the parcels list from the provided parcels.py file.

from parcels import parcels

print("Desired Real Estate Calculator")
name = input("Enter your name")
state_code = input("Enter desired state code")
while True:
    try:
        living_area = int(input("Enter desired living area"))
        break
    except ValueError:
        print("Invalid Entry. Please enter a number")
while True:
    try:
        minimum_cost = int(input("Enter desired Minimum Cost"))
        break
    except ValueError:
        print("Invalid Entry. Please enter a number")
while True:
    try:
        maximum_cost = int(input("Enter desired Maximum Cost"))

        if maximum_cost < minimum_cost:
            print("Invalid Entry. Maximum Cost cannot be less than Minimum Cost. Re enter valid Maximum Cost")
            continue
        break
    except ValueError:
        print("Invalid Entry. Please enter a number")

campus_name = input("Enter desired elementary campus name")
```

```python
# Section 2: Search for matching properties
match_count = 0

# Create an empty list to store the addresses of matching properties
matching_addresses = []

# Loop through every property in the parcels list
for parcel in parcels:
    if (
        parcel['state_code'] == state_code and
            parcel['living_area'] >= living_area and
            parcel['market_value'] >= minimum_cost and
            parcel['market_value'] <= maximum_cost and
            parcel['CAMPNAME'] == campus_name):
        match_count += 1
        matching_addresses.append(parcel['street_address'])
        
```

```python
# Section 3: Display the search results
# Print the total number of properties that matched all criteria
print("Number of matching properties:", match_count)

# Add heading before displaying the matching addresses
print("Matching addresses:")

# Print each matching address one at a time
for address in matching_addresses:
    print(address)
```
